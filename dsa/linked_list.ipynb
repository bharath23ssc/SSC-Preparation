```python
class Node:
    def __init__(self,data):
        self.data=data
        self.ref=None

class LinkedList:
    def __init__(self):
        self.head=None
    # Insert at beginning
    def insert_at_begin(self,data):
        if self.head is None:
            self.head=data
        else:
            n=Node(data)
            n.ref=self.head
            self.head=n
    # Insert at end
    def insert_at_end(self,data):
        if self.head is None:
            return self.insert_at_begin(data)
        else:
            new_node=Node(data)
            n=self.head
            while n.ref is not None:
                n=n.ref
            new_node.ref=n.ref
            n.ref=new_node

    # Insert at specific position (1-based index)
    def insert_at_specific(self,data,n):
        if n!=0 or n!=1:
            new_node=Node(data)
            n=self.head
            count=0
            while n.ref is not None and count<n-1:
                count+=1
                n=n.ref
            new_node.ref=n.ref
            n.ref=new_node
    # Middle of LL
    def middle_ll(self,data):
        if self.head is None:
            return  "Print Linked list is empty"
        new_node=Node(data)
        slow,fast=self.head,self.head
        while slow and fast:
            slow=slow.ref
            fast=fast.ref.ref
        new_node.ref=slow.ref
        slow.ref=new_node

    #traversal ll
    def traversal(self):
        n=self.head
        while n.ref is not None:
            print(n.data,"====>")
            n=n.ref

    # Delete by value
    def delete_by_value(self,data):
        if self.head is None:
            return "Linked list is empty"
        if self.head.data==data:
            self.head=self.head.ref
        else:
            n=self.head
            while n.ref is not None:
                if n.data==data:
                    n.ref=n.ref.ref
                    return
                n=n.ref
    # Delete by position
    def delete_by_position(self,position):
        if self.head is None:
            return "Linkde list is empty"
        else:
            count=0
            n=self.head
            while n.ref is not None and position<count-1:
                n=n.ref
                count+=1
            n.ref=n.ref.ref
    
    # Count number of nodes
    def count_number_of_node(self):
        count=0
        if self.head is None:
            return count
        else:
            n=self.head
            
            while n.ref is not None:
                count+=1
                n=n.ref
            return count
        
    # Search an element
    def search_element(self,data):
        if self.head is None:
            return
        else:
            n=self.head
            while n.ref is not None:
                if n.data==data:
                    return True
                n=n.ref
            return False

        


    
    
```
